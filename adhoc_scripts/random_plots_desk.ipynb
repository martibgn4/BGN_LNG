```python

import matplotlib.pyplot as plt
import numpy as np

# Example data (replace with yours)
years = np.array([2026, 2027, 2028, 2029, 2030])
base  = np.array([50, 55, 67.5, 77.5, 90])

# Lower/upper uncertainty bounds (replace with yours)
lower = np.array([45, 50, 60, 70, 80])
upper = np.array([55, 60, 75, 85, 100])

# Plot
plt.figure(figsize=(5, 4))

# Shaded uncertainty band
plt.fill_between(years, lower, upper, color="tab:blue", alpha=0.25, linewidth=0)

# Base line
plt.plot(years, base, color="tab:blue", linewidth=2, label="Base")

# Formatting to match style
plt.title("Spot [k$/day] (Uncertainty Bands)", fontsize=12)
plt.xlabel("Year", fontsize=10)
plt.ylabel("Freight Cost", fontsize=12)
plt.xticks(years)
plt.legend(loc="upper left")
plt.grid(False)

plt.tight_layout()
plt.show()
```
